```python
# importing of libaries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# loading the dataset
df = pd.read_csv("educational-facilities-in-nigeria.csv", encoding="latin-1")
df.head(10)
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
# identifying of missing values
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["management"].unique()
```

```python
# handling of inconsistence values
df["management"].replace({
    "faith_based": np.nan,
    "none": np.nan
}, inplace=True)
```

```python
# Handling of missing values
cat_cols = df.select_dtypes(include=["object", "float64"]).columns
df = df[cat_cols].dropna()
df
```

```python
# converting date to datetime
df["date_of_survey"] = pd.to_datetime(df["date_of_survey"], errors="coerce")
df["date_of_survey"]
```

```python
df["unique_lga"].unique()
```

```python
# Standardizing the Lga column
def standardize_lga(lga):
    if not isinstance(lga, str):
        return lga
    return lga.split("_")[0].split()[0].lower()
```

```python
df["unique_lga"] = df["unique_lga"].apply(standardize_lga)
df["unique_lga"]
```

```python
# Dropping of inconsistence values
df = df[df["facility_type_display"] != "Information not available / Don't know"]
```

```python
 df["facility_type_display"].unique()
```

```python
# Standardizing the facility_type_display column
df["facility_type_display"]= df["facility_type_display"].replace({
      "Primary Only":"Primary school only",
      "Pre-primary and Primary":"Pre-primary and Primary school combined",
      "Junior and Senior Secondary":"Junior and Senior Secondary school combined",
      "Primary and Junior Secondary":"Primary and Junior Secondary school combined",
      "Primary, Junior, and Senior Secondary":"Primary, Junior and Senior Secondary school combined",
      "Junior Secondary Only":"Junior Secondary school only",
      "Senior Secondary Only": "Senior Secondary school only",
      "Pre-primary Only": "Pre-primary school only"
  })
df["facility_type_display"].unique()
```

```python
# fixing the data type
df = df.convert_dtypes()
df.info()
```

```python
# capitalizing the string column
str_cols = df.select_dtypes(include='object')
standardized = str_cols.apply(lambda col: col.str.title())
```

```python
# dropping of unnecessary column
df = df.drop(columns=["formhub_photo_id", "survey_id", "gps", "facility_id"])
df
```

```python
# adding of total_teacher column
df["total_teacher"]  = df["num_tchrs_female"] + df["num_tchrs_male"]
df["total_teacher"]
```

```python
# adding of student teacher ratio column
df["student_teacher_ratio"] = df["num_students_total"] / df["total_teacher"]
df["student_teacher_ratio"]
```

```python
# adiing of total teacher column
df["total_student"] = df["num_students_female"] + df["num_students_male"]
df["total_student"]
```

```python
# converting the cleaned data to csv
df.to_csv("cleaned_education_dataset.csv", index=False)
```

```python
# common types of educational facilities
common_facilities = df["facility_type_display"].value_counts()
common_facilities
```

```python
# percentage of schools that has water suppply, sanitation,electriciy
water_pct = df["improved_water_supply"].mean().round(2) * 100
sanitation_pct = df["improved_sanitation"].mean().round(2) * 100
electricity_pct = df["phcn_electricity"].mean().round(2) * 100

print("Improved water supply: ", water_pct, "%")
print("Improved sanitation: ", sanitation_pct, "%")
print("Electricity: ", electricity_pct, "%")
```

```python
# average number of student per school
avg_students = df["total_student"].mean().round()
print("Average number of student per school:",avg_students)
```

```python
# average number of teacher per school
avg_teachers =df["total_teacher"].mean().round()
print(" average number of teacher per school:",avg_teachers)
```

```python
# comparasation of private and public schools in term of student population
Student_population = df.groupby("management")["total_student"].sum().sort_values(ascending=False)
Student_population
```

```python
# comparasation of private and public schools in term of teachers count
Teacher_count =  df.groupby("management")["total_teacher"].sum().sort_values(ascending=False)
Teacher_count
```

```python
# comparasation of private and public schools in term of infrastructure availability
Infrastructure_availability = df.groupby("management")[["improved_water_supply","improved_sanitation",  "phcn_electricity"]].mean() * 100
Infrastructure_availability
```

```python
# LGAs with the highest numbers of schools
df.groupby("unique_lga").size().sort_values(ascending=False).head(20)
```

```python
# LGAs with the lowest access to electricity
df.groupby("unique_lga")["phcn_electricity"].mean().sort_values().head(20)
```

```python
 # LGAs with the lowest access to sanitation
df.groupby("unique_lga")["improved_sanitation"].mean().sort_values(ascending=True).head(20)
```

```python
# Bar chart of school types
sns.countplot(data=df, x="management")
plt.title("Number of Schools by Type")
plt.xlabel("School Type")
plt.ylabel("Count")
plt.grid(axis="y",linestyle="--", alpha=0.7)
plt.show()
```
