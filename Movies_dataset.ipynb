```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
movies_df = pd.read_csv("movies.csv")
movies_df.head(10)
```

```python
movies_df.info()
```

```python
movies_df.shape
```

```python
movies_df.isnull().sum()
```

```python
movies_df['YEAR'] = movies_df["YEAR"].astype(str)
movies_df["YEAR"].tolist()
```

```python
def clean_year(value):
    if "("in value or ")" in value:
        value = value.replace("(", "").replace(")", "")
    if "-" in value or "-" in value:
        value = value.split("-")[0].split("-")[0]
    digit = "".join([i for i in value if i.isdigit()])
    if digit:
        return int(digit)
    else:
        return np.nan
movies_df["YEAR"] = movies_df["YEAR"].apply(clean_year)
movies_df["YEAR"]
```

```python
movies_df['YEAR'] = pd.to_numeric(movies_df['YEAR'], errors='coerce')

filtered_df = movies_df[(movies_df['YEAR'] >= 2010) & (movies_df['YEAR'] <= 2022)]
filtered_df
```

```python
movies_df.dropna(subset=["YEAR"], inplace=True)
```

```python
movies_df['GENRE'] = movies_df['GENRE'].astype(str)

movies_df['GENRE'].replace({
    "nan": np.nan,
    "\n": ""
}, inplace=True)

movies_df['GENRE'] = movies_df['GENRE'].dropna().apply(lambda x: [g.strip() for g in x.split(',')])

movies_df = movies_df.explode('GENRE')

movies_df = movies_df.dropna(subset=['GENRE']).reset_index(drop=True)

movies_df.head(10)
```

```python
movies_df["RATING"]= movies_df["RATING"].replace({"NaN": np.nan})
movies_df["RATING"]
```

```python
movies_df['RATING'] = pd.to_numeric(movies_df['RATING'], errors='coerce')
movies_df["RATING"]= movies_df['RATING'].fillna(movies_df['RATING'].mean()).round(2)
movies_df["RATING"]
```

```python
movies_df["VOTES"]
```

```python
movies_df["VOTES"].replace({"NaN": np.nan}, inplace=True)
```

```python
movies_df['VOTES'] = movies_df['VOTES'].astype(str).str.replace(r',', '', regex=False)

movies_df["VOTES"]
```

```python
movies_df['VOTES'] = pd.to_numeric(movies_df['VOTES'], errors='coerce')
movies_df['VOTES'].fillna(movies_df["VOTES"].mode()[0],inplace=True)
movies_df["VOTES"].tolist()
```

```python
movies_df["RunTime"].tolist()
movies_df["RunTime"] = movies_df["RunTime"].replace({"NaN":np.nan})
```

```python
movies_df["RunTime"]= movies_df["RunTime"].fillna(movies_df["RunTime"].mean()).round(1)
movies_df["RunTime"]
```

```python
movies_df["Gross"].tolist()
```

```python
movies_df["Gross"]= movies_df["Gross"].replace(r"[$M]" ,"", regex= True)
movies_df["Gross"].tolist()
```

```python
movies_df["Gross"]= movies_df["Gross"].replace({"NaN": np.nan})
```

```python
movies_df["Gross"] = pd.to_numeric(movies_df["Gross"], errors="coerce")
```

```python
movies_df["Gross"] = movies_df["Gross"].fillna(movies_df["Gross"].mean()).round(2)
movies_df["Gross"]
```

```python

movies_df.rename(columns={
    "Gross": "GROSS(Million)",
    "RunTime":"RUNTIME(Minutes)"
}, inplace=True)
movies_df
```

```python
movies_df['Director'] = movies_df["STARS"].str.extract(r'Director:\s*([^|]+)')
movies_df['Stars'] = movies_df["STARS"].str.extract(r'Stars:\s*(.*)')
movies_df['Director'] = movies_df['Director'].str.strip()
movies_df['Stars'] = movies_df['Stars'].str.strip()
    
movies_df
```

```python
movies_df.duplicated().sum()
```

```python
movies_df = movies_df.drop_duplicates()
```

```python
movies_df= movies_df.dropna(subset=["Director"], inplace=True)
```
