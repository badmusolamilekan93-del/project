```python
# 1. Group by customer and count their rentals
customer_counts = df.groupby('customer_id').size()

# 2. Identify repeat customers (rentals > 1)
repeat_customers = customer_counts[customer_counts > 1]

# 3. Calculate percentage
total_customers = len(customer_counts)
repeat_count = len(repeat_customers)
repeat_percentage = (repeat_count / total_customers) * 100

print(f"Total Customers: {total_customers}")
print(f"Repeat Customers: {repeat_count}")
print(f"Repeat Rental Rate: {repeat_percentage:.2f}%")
```

```python
# Total revenue divided by number of unique customers
avg_spending = df['total_amount_spent'].sum() / df['customer_id'].nunique()

print(f"Average Spending per Customer: ${avg_spending:.2f}")
```

```python
# 1. Ensure date is in datetime format
# 2. Sort and calculate the difference between rows for each customer
# df = df.sort_values(by=['customer_id', 'first_rental_date'])
# df['prev_rental'] = df.groupby('customer_id')['recent_rental_date'].shift(1)
df['time_gap'] = (df['recent_rental_date'] - df['first_rental_date']).dt.days

# 3. Calculate the mean gap (excluding the first rental which will be NaN)
avg_gap = df['time_gap'].mean()

print(f"Average Time Gap: {avg_gap:.1f} days")
```

```python
# 1. Find the last rental date per customer
last_rental = df['first_rental_date']
current_date = df['recent_rental_date'] # Using the latest date in dataset as "today"

recency = (current_date - last_rental).dt.days
churn_risk = recency[recency > 60 ].count()
inactive = recency[recency > 120].count()
active = recency[recency <= 60].count()

print(f"Active Customers: {active}")
print(f"At Risk (60+ days): {churn_risk}")
print(f"Inactive/Churned (120+ days): {inactive}")
```

```python
categories = ['Active', 'At Risk', 'Inactive']
counts = [active, churn_risk - inactive, inactive]

plt.figure(figsize=(8, 5))
plt.bar(categories, counts, color=['green', 'orange', 'red'])
plt.title('Customer Retention Status')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#  # films that generate the most revenue per rental
film_rev = df.groupby('film_title').agg(
    total_revenue=('film_total_rental_revenue', 'sum'),
    rental_count=('film_total_rental_revenue', 'count')
)

film_rev['rev_per_rental'] = film_rev['total_revenue'] / film_rev['rental_count']
top_10_films = film_rev.sort_values(by='rev_per_rental', ascending=False).head(10)
print("Top 10 Films by Revenue per Rental:")
print(top_10_films[['rev_per_rental', 'total_revenue', 'rental_count']])
```

```python
# categories that have the highest average rental price?
cat_price = df.groupby('category')['total_rental_revenue'].mean().sort_values(ascending=False)
cat_price_df = cat_price.reset_index()
print("Top 5 Most Expensive Categories:")
print(cat_price.head())
```

```python

film_stats = df.groupby('film_title').agg({
    'avg_rental_duration': 'first',
    'inventory_count': 'unique'
})

correlation = film_stats['avg_rental_duration'].corr(film_stats['inventory_count'])
print(f"Correlation between length and frequency: {correlation:.2f}")
```

```python
# Logic: Calculate rentals per available inventory item
inventory_demand = df.groupby('film_title').agg({
    'total_rental': 'nunique', 
    'inventory_count': 'nunique'
})

inventory_demand.columns = ['total_rental', 'stock_count']
inventory_demand['turnover_ratio'] = inventory_demand['total_rental'] / inventory_demand['stock_count']

high_demand_low_stock = inventory_demand[inventory_demand['total_rental'] > 10].sort_values('turnover_ratio', ascending=False)

print("Films with high demand but low inventory (Top 10):")
print(high_demand_low_stock.head(10))
```

```python

store_performance = df5.groupby('store_id')['total_revenue'].sum().sort_values(ascending=False)
total_revenue = store_performance.sum()
store_share = (store_performance / total_revenue) * 100

print("Store Profitability (Total Revenue):")
print(store_performance)
print("\nRevenue Share Percentage:")
print(store_share)
```

```python
# Which staff member is the most efficient
staff_efficiency = df5.groupby('full_name').agg(
    total_revenue=('total_revenue', 'sum'),
    total_rentals=('total_rent', 'unique')
)
staff_efficiency['avg_per_transaction'] = staff_efficiency['total_revenue'] / staff_efficiency['total_rentals']
staff_efficiency = staff_efficiency.sort_values(by='total_revenue', ascending=False)
print("Staff Efficiency Metrics:")
print(staff_efficiency)
```

```python
# 1. Aggregate total performance per staff member
staff_performance = df5.groupby('full_name').agg(
    total_revenue=('total_revenue', 'sum'),
    rental_count=('total_rent', 'unique')
).reset_index()

# 2. Calculate the Correlation Coefficient
correlation = staff_performance['rental_count'].corr(staff_performance['total_revenue'])

staff_performance['yield'] = staff_performance['total_revenue'] / staff_performance['rental_count']

print(f"Correlation between Volume and Revenue: {correlation:.2f}")
print("\nStaff Performance Comparison:")
print(staff_performance.sort_values(by='yield', ascending=False))
```

```python
regional_preferences = df.groupby(['country','city', 'category'])['total_amount_spent'].sum().reset_index()

regional_preferences = regional_preferences.sort_values(['country','city', 'total_amount_spent'], ascending=[True, True,False])
print("Country and Category Performance:")
print(regional_preferences.head(20))
```

```python
growth_metrics = df.groupby(['country', 'city']).agg(
    total_revenue=('total_amount_spent', 'sum'),
    unique_customers=('full_name', 'nunique'),
    total_rentals=('total_rental', 'nunique')
).reset_index()
growth_metrics['activity_rate'] = growth_metrics['total_rentals'] / growth_metrics['unique_customers']

# 3. Sort by Activity Rate to find the "Fastest" markets
fastest_regions = growth_metrics.sort_values(by='activity_rate', ascending=False).head(10)

print("Top 10 Regions with Highest Customer Activity :")
print(fastest_regions[['country', 'city', 'activity_rate', 'total_revenue']])
```
